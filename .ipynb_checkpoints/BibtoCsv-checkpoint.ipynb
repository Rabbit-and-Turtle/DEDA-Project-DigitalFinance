```python
import re
import pandas as pd
import bibtexparser

# ---------- 1. 读取 bib 文件（v2 API）----------
library = bibtexparser.parse_file("TOP5.bib")

records = []
for entry in library.entries:
    # 提取所有字段，并去掉 {} 花括号
    fields = {
        k: v.value.replace("{", "").replace("}", "").strip()
        for k, v in entry.fields_dict.items()
    }

    row = dict(fields)  # 所有原始字段全部保留
    row["citation_key"] = entry.key
    row["entry_type"] = entry.entry_type

    # ---------- 2. 解析 pages 字段，计算页数 ----------
    pages_raw = fields.get("pages", "")
    page_count = None
    if pages_raw:
        # 处理常见分隔符：--, –, —, +, ,
        parts = re.split(r"\s*(?:--+|–|—|\+|,)\s*", pages_raw)
        if len(parts) == 2:
            try:
                page_count = int(parts[1]) - int(parts[0]) + 1  # 闭区间
            except ValueError:
                pass
        elif len(parts) == 1 and parts[0].isdigit():
            page_count = 1

    row["page_count"] = page_count
    records.append(row)

# ---------- 3. 转成 DataFrame 并保存 ----------
df = pd.DataFrame(records)

# 列顺序：关键列在前，其余字段在后
front_cols = ["citation_key", "entry_type", "title", "author", "journal",
              "year", "volume", "number", "pages", "page_count",
              "doi", "url"]
front_cols = [c for c in front_cols if c in df.columns]
other_cols = [c for c in df.columns if c not in front_cols]
df = df[front_cols + other_cols]

df.to_csv("TOP5.csv", index=False, encoding="utf-8-sig")

print(df[["citation_key", "pages", "page_count"]])
print(f"\n共 {len(df)} 条记录，{len(df.columns)} 个字段，已保存到 items.csv")
```
