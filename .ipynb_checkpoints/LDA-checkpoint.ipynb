```python
# -*- coding: utf-8 -*-
"""
分年份 LDA 主题演化分析
  输入：CSV（abstract + year 两列，2019–2026）
  输出：
    1. 全局 LDA 模型 + 各主题关键词表
    2. 每年各主题的文档占比 → "主题 × 年份" 热力图
    3. 每年 Top 主题的词表 CSV
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
from wordcloud import STOPWORDS

# ============ 配置 ============
CSV_FILE   = "Data/digital_finance_aftercleaning.csv"
ABSTRACT_COL = "abstract"
YEAR_COL     = "year"

N_TOPICS    = 8       # 主题数 K，建议用 5/8/10/15/20 扫一遍再看
N_TOP_WORDS = 8       # 每主题展示的关键词数
OUT_HEATMAP = "topic_evolution_heatmap.png"
OUT_TOPICS  = "topic_words.csv"
OUT_SHARES  = "year_topic_shares.csv"

# ============ 1. 读入数据 ============
df = pd.read_csv(CSV_FILE)
df = df.dropna(subset=[ABSTRACT_COL, YEAR_COL])
df[YEAR_COL] = df[YEAR_COL].astype(int)
years = sorted(df[YEAR_COL].unique())
print(f"年份范围: {years[0]}–{years[-1]}，共 {len(df)} 篇摘要")


# ============ 领域泛词停用词表 ============
DOMAIN_STOPWORDS = [
    # ---- 学术写作套话（任何论文摘要都有，零信息量）----
    "paper", "study", "studies", "research", "using", "used", "use",
    "based", "propose", "proposed", "proposes", "article", "work",
    "result", "results", "finding", "findings", "conclusion", "conclusions",
    "analysis", "approach", "method", "methods", "model", "models",
    "framework", "evidence", "show", "shows", "shown", "examine",
    "examined", "investigate", "investigated", "explore", "explored",
    "aim", "aims", "purpose", "objective", "however", "also", "well",
    "many", "several", "various", "different", "significant",
    "important", "provide", "provides", "provide", "suggest",
    "suggests", "demonstrate", "demonstrates", "find", "finds",
    "present", "presented", "novel", "new", "existing", "recent",
    "literature", "empirical", "theoretical", "practical", "application",
    "applications", "effect", "effects", "impact", "impacts",
    "role", "factors", "level", "levels", "data", "sample", "samples",
    "test", "tests", "testing", "measure", "measures", "measured",
    "estimate", "estimates", "estimated", "variable", "variables",

    # ---- 领域泛词（数字金融文献人人都在说，不区分主题）----
    "financial", "finance", "digital", "market", "markets",
    "economy", "economic", "economics", "sector", "industry",
    "firm", "firms", "company", "companies", "business",
    "bank", "banks", "banking",          # 若"银行业"是你要的主题则移出
    "china", "chinese", "global", "world", "country", "countries",
    "development", "developing", "developed", "growth",
    "time", "period", "year", "years", "high", "low",
    "increase", "increased", "decrease", "change", "changes",
    "value", "values", "price", "prices",   # 视研究焦点决定
    "non","linear","short","long","term","standard","deviation","mean","variance","state","art","neural","networks","social","media"
]

# 应用：合并进现有 CountVectorizer


ALL_STOPWORDS = list(ENGLISH_STOP_WORDS) + DOMAIN_STOPWORDS


# ============ 2. 文档-词频矩阵（全局拟合，保证各年可比） ============
vectorizer = CountVectorizer(
    max_df=0.90,
    min_df=5,
    max_features=5000,
    stop_words=ALL_STOPWORDS,
    ngram_range=(2, 4),      # 保留 bigram，如 "digital finance"
    token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b",   # 只留 ≥2 字母的纯英文词
)

X = vectorizer.fit_transform(df[ABSTRACT_COL].astype(str))
print(f"词频矩阵: {X.shape[0]} 篇 × {X.shape[1]} 个词")

# ============ 3. 训练全局 LDA ============
lda = LatentDirichletAllocation(
    n_components=N_TOPICS,
    max_iter=20,
    learning_method="online",
    batch_size=1024,
    random_state=42,
)
doc_topic = lda.fit_transform(X)   # (n_docs, n_topics)，每行是文档的主题分布
print(f"困惑度: {lda.perplexity(X):.2f}")

# ============ 4. 主题关键词表 ============
feature_names = vectorizer.get_feature_names_out()
topic_words = {}
for k, topic in enumerate(lda.components_):
    top_idx = topic.argsort()[-N_TOP_WORDS:][::-1]
    topic_words[f"Topic {k}"] = [feature_names[i] for i in top_idx]

words_df = pd.DataFrame(topic_words)
words_df.to_csv(OUT_TOPICS, index=False, encoding="utf-8-sig")
print(f"\n主题关键词表已保存: {OUT_TOPICS}")
print(words_df.to_string())

# ============ 5. 各年主题占比 ============
df["dominant_topic"] = doc_topic.argmax(axis=1)

# 占比口径 A：每篇文档的主题分布概率求和后归一化（更平滑，推荐）
shares_prob = (
    pd.DataFrame(doc_topic, columns=[f"Topic {k}" for k in range(N_TOPICS)])
    .assign(year=df[YEAR_COL].values)
    .groupby("year").mean()
)
# 占比口径 B：每篇取主导主题再计比例（更直观）
shares_dom = (
    df.groupby([YEAR_COL, "dominant_topic"]).size().unstack(fill_value=0)
    .reindex(columns=range(N_TOPICS), fill_value=0)
)
shares_dom = shares_dom.div(shares_dom.sum(axis=1), axis=0)
shares_dom.columns = [f"Topic {k}" for k in range(N_TOPICS)]

shares_prob.to_csv(OUT_SHARES, encoding="utf-8-sig")

# ============ 6. 绘制演化热力图 ============
sns.set_theme(font="SimHei", style="whitegrid")   # 中文字体，按系统调整

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

for ax, shares, title in [
    (axes[0], shares_prob, "概率口径（文档主题分布均值）"),
    (axes[1], shares_dom, "主导主题口径"),
]:
    sns.heatmap(
        shares.T, ax=ax, cmap="YlOrRd",
        annot=True, fmt=".2f", annot_kws={"size": 8},
        cbar_kws={"label": "占比"},
    )
    ax.set_title(title, fontsize=13)
    ax.set_xlabel("年份")
    ax.set_ylabel("")
    # y 轴标签替换为主题关键词
    ax.set_yticklabels(
        [f"T{k}: {' '.join(topic_words[f'Topic {k}'][:4])}"
         for k in range(N_TOPICS)],
        rotation=0, fontsize=8,
    )

plt.tight_layout()
plt.savefig(OUT_HEATMAP, dpi=200, bbox_inches="tight")
print(f"\n热力图已保存: {OUT_HEATMAP}")

# ============ 7. 扫描主题数 K（可选， Uncomment 后运行） ============
# for k in [5, 8, 10, 15, 20]:
#     lda_k = LatentDirichletAllocation(
#         n_components=k, max_iter=20,
#         learning_method="online", random_state=42)
#     lda_k.fit(X)
#     print(f"K={k:2d}  perplexity={lda_k.perplexity(X):.2f}")
```

```python
# -*- coding: utf-8 -*-
"""
数字金融摘要 BERTopic 主题演化分析
  输入：CSV（abstract + year）
  输出：
    1. 主题关键词表 topic_words_bertopic.csv
    2. 文档-主题分配 document_topics.csv
    3. 年度主题占比热力图 topics_evolution_bertopic.png
    4. 主题间距离可视化（交互式 HTML + 静态图）
    5. 层次化主题树 topics_hierarchy.png
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sentence_transformers import SentenceTransformer
from bertopic import BERTopic
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer
import os
os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"   # 国内镜像站

# 然后再 import
from sentence_transformers import SentenceTransformer

# ============ 配置 ============
CSV_FILE     = "Data/digital_finance_aftercleaning.csv"
ABSTRACT_COL = "abstract"
YEAR_COL     = "year"
MIN_TOPIC_SIZE = 25        # 一个主题至少多少篇文档（越小主题越多越细）
NR_TOPICS      = None      # None=自动；想要固定数量改为如 12（会强制降维合并）
SEED = 42

OUT_WORDS = "topic_words_bertopic.csv"
OUT_DOCS  = "document_topics.csv"
OUT_HEAT  = "topics_evolution_bertopic.png"
OUT_VIZ   = "topics_intertopic.html"
OUT_HIER  = "topics_hierarchy.png"

# ============ 1. 读入数据 ============
df = pd.read_csv(CSV_FILE)
df = df.dropna(subset=[ABSTRACT_COL, YEAR_COL])
df[ABSTRACT_COL] = df[ABSTRACT_COL].astype(str)
df = df[df[ABSTRACT_COL].str.len() > 100]      # 剔除过短/损坏的摘要
docs = df[ABSTRACT_COL].tolist()
timestamps = df[YEAR_COL].astype(int).tolist()
print(f"文档数: {len(docs)}，年份: {min(timestamps)}–{max(timestamps)}")

# ============ 2. 语义嵌入 ============
# 首次运行会自动下载模型（~90MB）；离线环境见文末"常见问题"
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = embedding_model.encode(docs, show_progress_bar=True,
                                    batch_size=64)

# ============ 3. 子模型配置 ============
# UMAP：先降 50 维再降到 5 维（聚类友好），random_state 固定保证可复现
umap_model = UMAP(n_neighbors=15, n_components=5, min_dist=0.0,
                  metric="cosine", random_state=SEED)

# HDBSCAN：密度聚类，allow_single_document 防止全部被标为离群点
hdbscan_model = HDBSCAN(min_cluster_size=MIN_TOPIC_SIZE,
                        metric="euclidean",
                        cluster_selection_method="eom",
                        prediction_data=True)

# c-TF-IDF 的停用词：泛词在这里剔除，主题词表才会干净
domain_stopwords = [
    "paper", "study", "studies", "research", "using", "used", "use",
    "based", "propose", "proposed", "article", "result", "results",
    "finding", "findings", "show", "shows", "shown", "examine",
    "examined", "investigate", "explore", "aim", "however", "also",
    "well", "many", "several", "various", "different", "significant",
    "important", "provide", "suggest", "demonstrate", "find", "present",
    "novel", "new", "existing", "recent", "literature", "empirical",
    "analysis", "approach", "method", "methods", "model", "models",
    "framework", "evidence", "effect", "effects", "impact", "impacts",
    "data", "sample", "test", "measure", "level", "high", "low",
    "financial", "finance", "digital", "market", "markets",
    "economic", "economy", "sector", "industry", "firm", "firms",
    "china", "chinese", "global", "country", "countries", "development",
    "year", "years", "time", "value", "change", "increase",
]
vectorizer_model = CountVectorizer(
    stop_words=domain_stopwords,
    ngram_range=(1, 2),
    min_df=5,
    max_features=5000,
)

# ============ 4. 训练 BERTopic ============
topic_model = BERTopic(
    embedding_model=embedding_model,
    umap_model=umap_model,
    hdbscan_model=hdbscan_model,
    vectorizer_model=vectorizer_model,
    nr_topics=NR_TOPICS,              # None=自动定主题数
    calculate_probabilities=False,    # True 会慢很多，一般不需要
    verbose=True,
)
topics, probs = topic_model.fit_transform(docs, embeddings)

print(f"\n自动发现主题数: {len(topic_model.get_topics()) - 1}")  # -1 是离群点
print(f"离群点文档数: {sum(1 for t in topics if t == -1)}"
      f" ({sum(1 for t in topics if t == -1)/len(docs):.1%})")

# ============ 5. 主题词表 ============
topic_info = topic_model.get_topic_info()
print(topic_info[["Topic", "Count", "Name"]].head(15).to_string())

rows = []
for t in topic_info["Topic"]:
    if t == -1:
        continue
    words = [w for w, _ in topic_model.get_topic(t)[:10]]
    rows.append({"Topic": t, "Count": topic_info.loc[topic_info.Topic == t, "Count"].iloc[0],
                 "Top10": ", ".join(words)})
words_df = pd.DataFrame(rows).sort_values("Count", ascending=False)
words_df.to_csv(OUT_WORDS, index=False, encoding="utf-8-sig")
print(f"\n主题词表已保存: {OUT_WORDS}")

# ============ 6. 文档-主题分配（含离群点） ============
df_out = df.copy()
df_out["topic"] = topics
df_out.to_csv(OUT_DOCS, index=False, encoding="utf-8-sig")

# ============ 7. 年度主题演化 ============
# BERTopic 自带：按时间片统计每个主题的频率
topics_over_time = topic_model.topics_over_time(
    docs, timestamps,
    global_tuning=True,       # 用全局信息平滑各时间片
    evolution_tuning=True,
    datetime_format="%Y",
    nr_bins=len(set(timestamps)),   # 每年一个 bin
)

# 转成 "年份 × 主题占比" 矩阵画热力图
tot = topics_over_time.copy()
tot["year"] = tot["Timestamp"].astype(int)
pivot = tot.pivot_table(index="Topic", columns="year",
                        values="Frequency", aggfunc="sum")
pivot = pivot.div(pivot.sum(axis=0), axis=1)      # 每年归一化为占比
pivot = pivot.drop(index=-1, errors="ignore")     # 去掉离群点行

# 主题标签 = 该主题前 3 个词
label_map = {t: f"T{t}: " + " ".join(w for w, _ in topic_model.get_topic(t)[:3])
             for t in pivot.index}
pivot.index = [label_map.get(t, f"T{t}") for t in pivot.index]

plt.figure(figsize=(16, max(6, 0.5 * len(pivot))))
sns.heatmap(pivot, cmap="YlOrRd", annot=True, fmt=".2f",
            annot_kws={"size": 8}, cbar_kws={"label": "年度占比"})
plt.title("BERTopic 主题年度演化（数字金融摘要）", fontsize=14)
plt.xlabel("年份"); plt.ylabel("")
plt.tight_layout()
plt.savefig(OUT_HEAT, dpi=200, bbox_inches="tight")
print(f"演化热力图已保存: {OUT_HEAT}")

# ============ 8. 交互式主题间距离图（浏览器打开） ============
fig = topic_model.visualize_topics()
fig.write_html(OUT_VIZ)
print(f"交互式可视化已保存: {OUT_VIZ}（用浏览器打开）")

# ============ 9. 层次化主题树（看主题合并关系，选主题数很有用） ============
hier_topics, tree = topic_model.hierarchical_topics(docs)
fig_h = topic_model.visualize_hierarchy()
fig_h.write_html("topics_hierarchy.html")
fig_h.write_image(OUT_HIER, width=1400, height=900)  # 需要 kaleido: pip install kaleido
print("层次结构已保存: topics_hierarchy.html / topics_hierarchy.png")

# ============ 10. 人工解读：抽读每个主题的代表作 ============
for t in words_df["Topic"].head(6):     # 只抽前 6 大主题
    print(f"\n===== 主题 {t} | {label_map.get(t)} =====")
    reps = topic_model.get_representative_docs(t)
    for i, doc in enumerate(reps[:2]):
        print(f"[文档{i+1}] {doc[:200]}...")
```
